```python
import pandas as pd
from IPython.display import display

df_stations = pd.read_csv('data/Citycycle_stations.csv')
df_trajets = pd.read_csv('data/Citycycle_trajets.csv')
df_abonnees = pd.read_csv('data/Citycycle_abonnes.csv')
```

```python
print("="*50)
print("EXPLORATION DU FICHIER TRAJETS")
print("="*50)
print("Dimensions :", df_trajets.shape)
display(df_trajets.head(7))
df_trajets.info()
display(df_trajets.isna().sum())
display(df_trajets.describe())
```
